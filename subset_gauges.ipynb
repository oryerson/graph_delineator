Subset all_matchups.csv
allows filter of the `all_matchups.csv` file by a specific `outlet_id` and save the result as a new CSV file.

```python
import pandas as pd

# Load the data
df = pd.read_csv("all_matchups.csv")

# Subset to the Columbia River basin
columbia_basin = df[df['outlet_id'] == 'USGS-14246900']

# Save the subset to a new CSV
columbia_basin.to_csv("columbia_basin_gauges.csv", index=False)
```
